```python
# Uwaga: wymiary muszą być w podwójnym nawiasie jako krotka, np. ((5, 5))
ones = np.ones((5, 5))
zeros = np.zeros((3, 6))
print("Ones:\n", ones)
print("Zeros:\n", zeros)
```

```python
rand_arr_6x3 = np.random.rand(6, 3)
rand_arr_trans = rand_arr_6x3.T

print("Original (6x3):\n", rand_arr_6x3)
print("\nTransposed (3x6):\n", rand_arr_trans)
```

```python
# Definiujemy przykładowe wymiary n, p, m
n, p, m = 2, 3, 4

arr_a = np.random.rand(n, p) # macierz 2x3
arr_b = np.random.rand(p, m) # macierz 3x4

# Iloczyn skalarny / mnożenie macierzy (dot product)
dot_product = np.dot(arr_a, arr_b) # alternatywnie: arr_a @ arr_b

print("Array A (2x3):\n", arr_a)
print("\nArray B (3x4):\n", arr_b)
print("\nDot Product (2x4):\n", dot_product)
```
